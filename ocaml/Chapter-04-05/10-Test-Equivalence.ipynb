**Function `test sigma s1 s2`**
- *Input:* `sigma` is an alphabet and `s1` and `s2` are strings that represent regular expressions.
- *Output:* `()`.  The function prints whether the two regular expressions are equivalent.

```ocaml
let test (sigma : char FSet.t) (s1 : string) (s2 : string) : unit =
  let r1 = parse s1 in
  let r2 = parse s2 in
  if regExpEquiv r1 r2 sigma then
    Printf.printf "The regular expressions %s and %s are equivalent.\n%!" s1 s2
  else
    Printf.printf "The regular expressions %s and %s are not equivalent.\n%!" s1 s2
```

```ocaml
let sigma = FSet.of_list [ 'a'; 'b'; 'c' ]
```

```ocaml
test sigma "a*" "a**"
```

The symbol `𝜀` can be created using the $\LaTeX$ expression `\varepsilon`: $\varepsilon$

The symbol `∅` can be created using the $\LaTeX$ expression `\emptyset`: $\emptyset$

```ocaml
test sigma "∅*" "𝜀"
```

```ocaml
test sigma "𝜀*" "𝜀"
```

```ocaml
test sigma "(𝜀+a)(a+𝜀)*(a+𝜀)" "a*"
```

```ocaml
test sigma "(𝜀+a)*" "a*"
```

```ocaml
test sigma "(ba)(ba)*" "b(ab)*a"
```

```ocaml
test sigma "𝜀" "∅**"
```

```ocaml
test sigma "(a+b+c)*(ac*b+bc*a)(a+b+c)*" "c*(a(a+c)*b+b(b+c)*a)(a+b+c)*"
```

```ocaml
test sigma "(a+b+c)*(ac*b+bc*a)(a+b+c)*" "c*(a(a+c)*b+b(b+c)*a)c*"
```

```ocaml
let sigma = FSet.of_list [ '0'; '1' ]
```

```ocaml
test sigma "1*(0+11+111)*1*" "1*((111*)*0*)*1*"
```
