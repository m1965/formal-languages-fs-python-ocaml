# Test the Equivalence of Regular Expressions

```
// Checks remain active in both F# Interactive and the Jupyter kernel.
let checkNotebook condition =
    if not condition then failwith "Notebook check failed"
```

```
open AutomataTypes
open RegExpParser
open RegExpToNfa
open NfaToDfa
open Minimize
open Equivalence
```

```
let test alphabet first second expected =
    let equivalent = regExpEquiv (parse first) (parse second) alphabet
    printfn "The regular expressions %s and %s are %sequivalent." first second (if equivalent then "" else "not ")
    checkNotebook (equivalent = expected)
let alphabet = set ['a';'b';'c']
```

The symbol `𝜀` can be created using the $\LaTeX$ expression `\varepsilon`: $\varepsilon$

The symbol `∅` can be created using the $\LaTeX$ expression `\emptyset`: $\emptyset$

```
test alphabet "a*" "a**" true
```

```
test alphabet "∅*" "𝜀" true
```

```
test alphabet "𝜀*" "𝜀" true
```

```
test alphabet "(𝜀+a)(a+𝜀)*(a+𝜀)" "a*" true
```

```
test alphabet "(𝜀+a)*" "a*" true
```

```
test alphabet "(ba)(ba)*" "b(ab)*a" true
```

```
test alphabet "𝜀" "∅**" true
```

```
test alphabet "(a+b+c)*(ac*b+bc*a)(a+b+c)*" "c*(a(a+c)*b+b(b+c)*a)(a+b+c)*" true
```

```
test alphabet "(a+b+c)*(ac*b+bc*a)(a+b+c)*" "c*(a(a+c)*b+b(b+c)*a)c*" false
```

The final pair is intentionally inequivalent: the first language accepts `abba` but the second does not. These examples also distinguish the empty language from the language containing only the empty word.

```
test alphabet "∅" "𝜀" false
test alphabet "a+b" "b+a" true
```
