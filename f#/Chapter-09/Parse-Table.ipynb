# A Parse Table for a Shift-Reduce Parser

This notebook contains the parse table that is needed for a *shift reduce parser* that parses the following grammar:
$$
  \begin{eqnarray*}
  \mathrm{expr}        & \rightarrow & \mathrm{expr}\;\;\texttt{'+'}\;\;\mathrm{product}   \\
                       & \mid        & \mathrm{expr}\;\;\texttt{'-'}\;\;\mathrm{product}   \\
                       & \mid        & \mathrm{product}                                    \\[0.2cm]
  \mathrm{product}     & \rightarrow & \mathrm{product}\;\;\texttt{'*'}\;\;\mathrm{factor} \\
                       & \mid        & \mathrm{product}\;\;\texttt{'/'}\;\;\mathrm{factor} \\
                       & \mid        & \mathrm{factor}                                     \\[0.2cm]
  \mathrm{factor}      & \rightarrow & \texttt{'('} \;\;\mathrm{expr} \;\;\texttt{')'}     \\
                       & \mid        & \texttt{NUMBER} 
  \end{eqnarray*}
$$

Below, we define the *grammar rules*.

```
let r1 = { Head = "e"; Body = [Nonterminal "e"; Terminal "+"; Nonterminal "p"] }
let r2 = { Head = "e"; Body = [Nonterminal "e"; Terminal "-"; Nonterminal "p"] }
let r3 = { Head = "e"; Body = [Nonterminal "p"] }
let r4 = { Head = "p"; Body = [Nonterminal "p"; Terminal "*"; Nonterminal "f"] }
let r5 = { Head = "p"; Body = [Nonterminal "p"; Terminal "/"; Nonterminal "f"] }
let r6 = { Head = "p"; Body = [Nonterminal "f"] }
let r7 = { Head = "f"; Body = [Terminal "("; Nonterminal "e"; Terminal ")"] }
let r8 = { Head = "f"; Body = [Terminal "NUMBER"] }
```

Below is the definition of the *goto table*.

```
let gotoTable = Map.ofList [
    (0, "e"), 4
    (0, "p"), 3
    (0, "f"), 1
    (5, "e"), 6
    (5, "p"), 3
    (5, "f"), 1
    (8, "p"), 15
    (8, "f"), 1
    (9, "p"), 10
    (9, "f"), 1
    (11, "f"), 14
    (12, "f"), 13
]
```

Finally, we define the *state table*.  This is table is only used for pretty printing.  This table gives us a clue about the information that is stored in the different stats.

```
// These descriptions are for teaching and trace output only.
let stateDescriptions = [|
    "ŝ -> • e; e -> • e + p; e -> • e - p; e -> • p; p -> • p * f; p -> • p / f; p -> • f; f -> • ( e ); f -> • NUMBER"
    "p -> f •"
    "f -> NUMBER •"
    "p -> p • * f; p -> p • / f; e -> p •"
    "ŝ -> e •; e -> e • + p; e -> e • - p"
    "f -> ( • e ); e -> • e + p; e -> • e - p; e -> • p; p -> • p * f; p -> • p / f; p -> • f; f -> • ( e ); f -> • NUMBER"
    "f -> ( e • ); e -> e • + p; e -> e • - p"
    "f -> ( e ) •"
    "e -> e + • p; p -> • p * f; p -> • p / f; p -> • f; f -> • ( e ); f -> • NUMBER"
    "e -> e - • p; p -> • p * f; p -> • p / f; p -> • f; f -> • ( e ); f -> • NUMBER"
    "e -> e - p •; p -> p • * f; p -> p • / f"
    "p -> p / • f; f -> • ( e ); f -> • NUMBER"
    "p -> p * • f; f -> • ( e ); f -> • NUMBER"
    "p -> p * f •"
    "p -> p / f •"
    "e -> e + p •; p -> p • * f; p -> p • / f"
|]
let table : ParseTable<string> =
    { States = stateDescriptions |> Array.map Set.singleton
      Actions = actionTable
      Gotos = gotoTable }
```
